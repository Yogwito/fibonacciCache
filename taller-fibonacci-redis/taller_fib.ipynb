## Solucion preguntas: 

6. Si ejecutas el programa otra vez dentro de los 30 segundos siguientes, ¿cambiaría el inventario? ¿Y si esperas más de 30 segundos sin ejecutar FLUSHDB?
Si ejecutamos el programa otra vez dentro de los 30 segundos, el inventario no cambia: siguen las mismas 6 claves, porque las fib:* aún no expiran y todas las consultas serían HIT. Solo cambian los valores (los contadores y el historial siguen aumentando), si esperamos más de 30 segundos sin ejecutar FLUSHDB, las claves fib:* expiran y se borran, y el inventario baja de 6 a 3 claves: quedan solo historial, stats:hits y stats:misses, que son permanentes. Al consultar de nuevo serían MISS, se recalcularía y las claves fib:* se volverían a crear.

## Preguntas que estan en el taller: 
- Parte 6: ¿por qué la lista y el set dan cantidades diferentes?
La lista permite repetir valores porque hace referencia a un historial, y un historial debe registrar cada consulta, aunque se repita, mientras que el set no guarda valores repetidos; por esta razón se dan cantidades diferentes.

- Parte 7: ¿fue HIT o MISS? ¿Por qué?
Fue MISS porque primero cambiamos el TTL de fib:30 a 1 segundo y luego detuvimos la ejecución 2 segundos con time.sleep. Pasado ese tiempo la clave expiró y se borró de Redis, por eso EXISTS devolvió 0 (la clave ya no existe). Al volver a llamar fib_cache(30), no la encontró en el caché, así que tuvo que recalcular el valor y lo guardó otra vez: eso es un MISS.

Preguntas 2, 3 y 4

¿Qué pasaría con el caché si no usaras TTL?

Sin el TTL las claves no expirarían nunca y por lo tanto se quedarían almacenadas, esto provocaría que se acumulen y ocupen espacio en memoria que luego se tendría que limpiar manualmente si se quiere guardar algo más.

3. ¿En qué casos reales usarías un caché con expiración?

En una página web se podrían guardar imagenes u otros elementos ya que es muy normal cerrar y entrar a una página repetidas veces, el caché ayudaría a que cargue más rapido cuando se vuelve a visitar. Tambien podría ser útil en videojuegos online para guardar información de partidas como las estadísticas y que no se tengan que consultar al servidor constantemente

**Bonus (Memoización): ¿Por qué `fib_memo(90)` es casi instantáneo, si con la versión de la parte 2 `fib(35)` ya tardaba tanto?**

En la versión sin caché (`fib`), cada llamada se divide en dos llamadas recursivas sin guardar resultados, recalculando los mismos subproblemas millones de veces con complejidad exponencial $O(2^n)$ (por eso `fib(35)` tardó 486.76 ms). En cambio, `fib_memo(90)` guarda cada subproblema `fib(k)` en Redis bajo `memo:<k>`: cada valor se calcula una sola vez y las llamadas repetidas se leen de Redis en tiempo constante $O(1)$, reduciendo la complejidad a lineal $O(n)$ (39.60 ms).

**Pregunta 5: Según el inventario de la parte 8, ¿qué estructura ocupa más claves y por qué?**

El prefijo que más claves ocupa es `memo:*` (tipo `string`), con **89 claves** (`memo:2` a `memo:90`) de un total de **96 claves** (`DBSIZE = 96`), porque `fib_memo(90)` crea una clave independiente por cada subproblema intermedio calculado. En comparación, `fib:*` solo guarda las 3 consultas distintas de la secuencia (`fib:25`, `fib:30`, `fib:32`), `stats:*` usa 2 claves (`stats:hits`, `stats:misses`), y la lista (`historial`) y el set (`consultados`) agrupan todos sus elementos en 1 sola clave cada uno.

**Tabla del inventario de claves (Parte 8):**

| Categoría | Elemento / Prefijo | Tipo (`TYPE`) | Cantidad | Expiración (`TTL`) |
|---|---|---|---|---|
| Total (`DBSIZE`) | Todas las claves en DB 1 | — | **96** | 3 con TTL / 93 permanentes |
| Por prefijo | `fib:*` (`fib:25`, `fib:30`, `fib:32`) | `string` | **3** | Con expiración (~26s–30s) |
| Por prefijo | `memo:*` (`memo:2` a `memo:90`) | `string` | **89** | Permanente (`TTL = -1`) |
| Por prefijo | `stats:*` (`stats:hits`, `stats:misses`) | `string` | **2** | Permanente (`TTL = -1`) |
| Por prefijo | Sin prefijo (`historial`) | `list` | **1** | Permanente (`TTL = -1`) |
| Por prefijo | Sin prefijo (`consultados`) | `set` | **1** | Permanente (`TTL = -1`) |

| Métrica final | Valor | Detalle |
|---|---|---|
| `string` / `list` / `set` | **94 / 1 / 1** | Total 96 claves |
| `stats:hits` / `stats:misses` | **4 / 4** | Tasa de aciertos final: **50.0%** (incluyendo Parte 7) |
